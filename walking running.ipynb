```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.utils import to_categorical
```

```python
df = pd.read_csv("walkrun.csv")

df = df.drop(columns=['date', 'time', 'username'])
```

```python
X = df.drop('activity', axis=1).values
y = df['activity'].values
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
def create_sequences(X, y, time_steps=20):
    Xs, ys = [], []
    for i in range(len(X) - time_steps):
        Xs.append(X[i:i+time_steps])
        ys.append(y[i+time_steps])
    return np.array(Xs), np.array(ys)

X_seq, y_seq = create_sequences(X_scaled, y, time_steps=20)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_seq, y_seq, test_size=0.2, random_state=42
)
```

```python
model = Sequential([
    LSTM(64, input_shape=(X_train.shape[1], X_train.shape[2])),
    Dropout(0.3),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.2
)
```

```python
loss, accuracy = model.evaluate(X_test, y_test)
print("LSTM Accuracy:", accuracy)
```

```python
sample = X_seq[0].reshape(1, X_seq.shape[1], X_seq.shape[2])

prediction = model.predict(sample)

print("Activity:",
      "Running 🏃" if prediction[0][0] > 0.5 else "Walking 🚶")
```

## Challenges Faced

During the implementation of the Human Activity Recognition (Walking vs Running) project, several challenges were encountered:

### 1. Raw Sensor Data Noise
The accelerometer and gyroscope sensors produced noisy signals due to sudden hand movements and sensor sensitivity. This affected model stability and required proper preprocessing and scaling.

### 2. Feature Selection
The dataset contained non-informative attributes such as date, time, and username. Identifying and removing these columns was necessary to improve model performance.

### 3. Data Imbalance Handling
Although the dataset contained both walking and running activities, ensuring balanced representation during training and testing was essential to avoid biased predictions.

### 4. Scaling of Sensor Values
Sensor readings varied in magnitude across different axes. Without normalization, machine learning models such as Logistic Regression and LSTM showed poor convergence.

### 5. Time-Series Sequence Creation for LSTM
Deep Learning models require sequential input data. Converting raw sensor readings into fixed-length time windows while preserving temporal information was a key challenge.

### 6. Model Selection and Overfitting
Choosing the optimal model involved comparing multiple algorithms. Tree-based models tended to overfit, while deep learning models required careful tuning of epochs and dropout layers.

### 7. Environment and Dependency Issues
Installing TensorFlow and managing compatible Python versions posed challenges, especially when multiple Python versions were present on the system.

### 8. Computational Complexity
Training LSTM models on large datasets increased computation time and memory usage, requiring optimization of batch size and number of epochs.

## Limitations

1. **Limited Activity Classes**  
   The current model only classifies walking and running, which restricts its applicability to a broader range of human activities.

2. **Sensor Noise Sensitivity**

3. **High Computational Cost for Deep Learning**  
   LSTM models require more computational resources and training time compared to traditional machine learning models.

4. **Dependency on Sensor Placement**

5. **Data Dependency**  
   Deep learning models require large volumes of labeled data to achieve optimal performance.

6. **Limited Generalization**  
   Models trained on specific users or devices may not generalize well to unseen users or different hardware without retraining.
