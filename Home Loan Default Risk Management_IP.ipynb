#  Home Loan Default Risk Management

##  1. Introduction

Home loan default is a major financial risk for banks and financial institutions.  
This project aims to build a **Machine Learning classification model** that predicts whether a customer is likely to default on a home loan.

Early prediction helps in:
- Reducing financial losses
- Improving loan approval decisions
- Maintaining a healthy credit portfolio

---

##  2. Problem Statement

Given customer details such as income, loan amount, credit history, employment status, and property area, predict:

> **Loan Status → Default (1) or Non-Default (0)**

This is a **Binary Classification Problem**.

---

##  3. Dataset Description

The dataset contains the following key features:

- ApplicantIncome  
- CoapplicantIncome  
- LoanAmount  
- Loan_Amount_Term  
- Credit_History  
- Property_Area  
- Loan_Status (Target Variable)

Target Variable:
- 1 → Default
- 0 → Non Default

---

##  4. Data Preprocessing

### 4.1 Handling Missing Values

- Numerical columns filled using Mean
- Categorical columns filled using Mode

### 4.2 Encoding Categorical Variables

- Label Encoding / One-Hot Encoding applied
- Converted categorical data into numerical format

### 4.3 Feature Scaling

- StandardScaler used to normalize feature values
- Improves model convergence and performance

---

##  5. Exploratory Data Analysis (EDA)

Key observations:

- Applicants with good credit history have lower default rate.
- Higher loan amounts slightly increase default probability.
- Income distribution shows majority applicants fall in middle-income category.

Visualization libraries used:
- Matplotlib
- Seaborn

```python
def cramers_v(x, y):
    confusion_matrix = pd.crosstab(x, y)
    chi2 = chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    phi2 = chi2 / n
    r, k = confusion_matrix.shape
    phi2corr = max(0, phi2 - ((k-1)*(r-1))/(n-1))
    rcorr = r - ((r-1)**2)/(n-1)
    kcorr = k - ((k-1)**2)/(n-1)
    return np.sqrt(phi2corr / min((kcorr-1), (rcorr-1)))
    

camers_results = {}
for col in categorical_cols:
    try:
        cv = cramers_v(Dulp2_final_data['TARGET'], Dulp2_final_data[col])
        camers_results[col] = cv
    except:
        camers_results[col] = np.nan
cramers_df = pd.DataFrame.from_dict(camers_results, orient='index', columns=['Cramers_V']).sort_values(by='Cramers_V', ascending=False) 
```

```python
numerical_data = Dulp2_final_data[numerical_cols].dropna(axis=1, how='any')
pearson_corr_matrix = numerical_data.corr().abs()



high_corr_pairs = (pearson_corr_matrix.where(np.triu(np.ones(pearson_corr_matrix.shape), k=1).astype(bool))
                   .stack()
                   .reset_index()
)
high_corr_pairs.columns = ['Feature_1', 'Feature_2', 'Correlation']
high_corr_pairs = high_corr_pairs[high_corr_pairs['Correlation'] > 0.9]


vif_data = pd.DataFrame()
vif_data['Feature'] = numerical_data.columns
vif_data['VIF'] = [variance_inflation_factor(numerical_data.values, i) for i in range(numerical_data.shape[1])]
high_vif = vif_data[vif_data['VIF'] > 10]


def cramers_v(x, y):
    confusion_matrix = pd.crosstab(x, y)
    chi2 = chi2_contingency(confusion_matrix, correction=False)[0]
    n = confusion_matrix.sum().sum()
    phi2 = chi2 / n
    r, k = confusion_matrix.shape
    phi2corr = max(0, phi2 - ((k-1)*(r-1))/(n-1))
    rcorr = r - ((r-1)**2)/(n-1)
    kcorr = k - ((k-1)**2)/(n-1)
    denom = min((kcorr-1), (rcorr-1))
    if denom <= 0:
        return np.nan
    else:
        return np.sqrt(phi2corr / denom)
def compute_cramers_v_matrix(df, categorical_cols):
    n = len(categorical_cols)
    cramers_v_matrix = pd.DataFrame(np.zeros((n, n)), index=categorical_cols, columns=categorical_cols)
    for i in range(n):
        for j in range(i, n):
            val = cramers_v(df[categorical_cols[i]], df[categorical_cols[j]])
            cramers_v_matrix.iloc[i, j] = val
            cramers_v_matrix.iloc[j, i] = val
    return cramers_v_matrix 

cramers_v_matrix = compute_cramers_v_matrix(Dulp2_final_data, categorical_cols)

print("\n highly correlated categorical feature pairs (Cramér's V > 0.9):")
print(high_corr_pairs)

print("\nFeatures with high VIF (>10):")
print(high_vif)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import OneHotEncoder, LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import (classification_report, confusion_matrix, roc_auc_score,
                             precision_recall_curve, auc, fbeta_score,
                             RocCurveDisplay, PrecisionRecallDisplay)

from sklearn.linear_model import LogisticRegression, RidgeClassifier
from sklearn.svm import LinearSVC
from sklearn.neighbors import KNeighborsClassifier
from imblearn.pipeline import Pipeline as ImbPipeline
import warnings
```

```python
df2_final_data = Dulp2_final_data.copy()
target = 'TARGET'

x = df2_final_data.drop(columns=[target])
y = df2_final_data[target].astype(int)


categorical_cols = x.select_dtypes(include=['object', 'category']).columns.tolist()
low_card = [col for col in categorical_cols if x[col].nunique() < 5]
high_card = list(set(categorical_cols) - set(low_card))

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)


x_train_label = pd.DataFrame(index=x_train.index)
x_test_label = pd.DataFrame(index=x_test.index)


for col in high_card:
    le = LabelEncoder()
    x_train_label[col] = le.fit_transform(x_train[col].astype(str))
    x_test_label[col] = x_test[col].astype(str).map(lambda x: le.transform([x])[0] if x in le.classes_ else -1)

ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False, drop='first')
x_train_ohe_array = ohe.fit_transform(x_train[low_card])
x_test_ohe_array = ohe.transform(x_test[low_card])


x_train_ohe = pd.DataFrame(x_train_ohe_array, 
                           columns=ohe.get_feature_names_out(low_card),
                             index=x_train.index)


x_test_ohe = pd.DataFrame(x_test_ohe_array, 
                          columns=ohe.get_feature_names_out(low_card),
                            index=x_test.index)





other_col = x.columns.difference(high_card + low_card)
x_train_other = x_train[other_col].copy()
x_test_other = x_test[other_col].copy()

for col in other_col:
    if x_train_other[col].dtype == 'object':
        unique_value = set(x_train_other[col].dropna().unique())
        if unique_value <= {'Y', 'N'}:
            x_train_other[col] = x_train_other[col].map({'Y': 1, 'N': 0})
            x_test_other[col] = x_test_other[col].map({'Y': 1, 'N': 0})
        else:
            le = LabelEncoder()
            x_train_other[col] = le.fit_transform(x_train_other[col].astype(str))
            x_test_other[col] = x_test_other[col].astype(str).map(lambda x: le.transform([x])[0] if x in le.classes_ else -1)
    


x_train = pd.concat([x_train_label, x_train_ohe, x_train_other], axis=1)
x_test = pd.concat([x_test_label, x_test_ohe, x_test_other], axis=1)


model = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42, class_weight='balanced'),
    'Ridge Classifier': RidgeClassifier(random_state=42, class_weight='balanced'),
    'Linear SVC': LinearSVC(max_iter=10000, random_state=42, class_weight='balanced'),
    'KNN': KNeighborsClassifier(n_neighbors=5)
}


all_model_results = []
local_results_1 = []


def evaluate_model(name, model, apply_smote=False):
    steps = [('scaler', StandardScaler())]
    if apply_smote:
        from imblearn.over_sampling import SMOTE
        steps.append(('smote', SMOTE(random_state=42)))
    steps.append(('classifier', model))

    pipeline = ImbPipeline(steps)
    pipeline.fit(x_train, y_train)

    if hasattr(pipeline.named_steps['classifier'], 'predict_proba'):
            proba = pipeline.predict_proba(x_test)[:, 1]
    else:
         try:
            proba = pipeline.decision_function(x_test)
            proba = (proba - proba.min()) / (proba.max() - proba.min())
         except:
            proba = pipeline.predict(x_test)

    thresholds = np.linspace(0.1, 0.9, 81)
    best_fbeta, best_thresh = 0, 0.3
    for thresh in thresholds:
        preds = (proba >= thresh).astype(int)
        fbeta = fbeta_score(y_test, preds, beta=2)
        if fbeta > best_fbeta:
            best_fbeta = fbeta
            best_thresh = thresh

    final_preds = (proba >= best_thresh).astype(int)

    report = classification_report(y_test, final_preds, output_dict=True)
    class_metrics = report['1']
    roc_auc = roc_auc_score(y_test, proba)
    precision, recall, _ = precision_recall_curve(y_test, proba)
    pr_auc = auc(recall, precision)

    result = {
        'Model': name,
        'best_threshold': best_thresh,
        'roc_auc': roc_auc,
        'pr_auc': pr_auc,
        'precision': class_metrics['precision'],
        'recall': class_metrics['recall'],
        'f1_score': class_metrics['f1-score'],
        'params': model.get_params() if hasattr(model, 'get_params') else str(model),
        'probs': proba,
        'preds': final_preds
    }
    local_results_1.append(result)

    print(f"\n==={name}===")
    print("Classification Report:")                
    print(classification_report(y_test, final_preds))
    print("Confusion Matrix:")
    print(confusion_matrix(y_test, final_preds))


    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    from sklearn.metrics import RocCurveDisplay, PrecisionRecallDisplay
    RocCurveDisplay.from_estimator(pipeline, x_test, y_test, ax=axes[0])
    axes[0].set_title(f'ROC Curve - {name}')
    axes[0].grid(True)

    PrecisionRecallDisplay.from_estimator(pipeline, x_test, y_test, ax=axes[1])
    axes[1].set_title(f'Precision-Recall Curve - {name}')
    axes[1].grid(True)

    plt.tight_layout()
    plt.show()



for name, model in model.items():
    evaluate_model(name, model)


best_model = max(local_results_1, key=lambda x: (x['roc_auc'] + x['f1_score'] + x['precision'] + x['recall']))
all_model_results.append(best_model)

print("\nBest Model Based on Combined Metrics:")
print(f"Model: {best_model['Model']}")
print(f"ROC AUC: {best_model['roc_auc']:.4f}")
print(f"Precision: {best_model['precision']:.4f}")
print(f"Recall: {best_model['recall']:.4f}")
print(f"F1 Score: {best_model['f1_score']:.4f}")
print(f"Best Threshold: {best_model['best_threshold']:.4f}")
print(f"Model Parameters: {best_model['params']}")


print("\n Classification Report for Best Model:")
print(classification_report(y_test, best_model['preds']))
print("\n Confusion Matrix for Best Model:")
print(confusion_matrix(y_test, best_model['preds']))



fig, axes = plt.subplots(1, 2, figsize=(12, 5))
RocCurveDisplay.from_predictions(y_test, best_model['probs'], ax=axes[0])
axes[0].set_title(f'ROC Curve - {best_model["Model"]}')
axes[0].grid(True)

PrecisionRecallDisplay.from_predictions(y_test, best_model['probs'], ax=axes[1])
axes[1].set_title(f'Precision-Recall Curve - {best_model["Model"]}')
axes[1].grid(True)


plt.tight_layout()
plt.show()
```

```python
from sklearn.base import clone
from sklearn.model_selection import RandomizedSearchCV
from sklearn.utils._testing import ignore_warnings
from sklearn.exceptions import ConvergenceWarning
from imblearn.over_sampling import SMOTE


@ignore_warnings(category=ConvergenceWarning)
def tune_and_evaluate(base_model_info, x_train, y_train, x_test, y_test, apply_smote = False):
    name = base_model_info['Model']
    model = base_model_info['Model_instance']
    print(f"\nTuning {name}...")

    if isinstance(model, LogisticRegression):
        param_dist = {
            'classifier__C': np.logspace(-4, 4, 20),
            'classifier__solver': ['liblinear', 'saga'],
            'classifier__penalty': ['l1', 'l2']
        }
    elif isinstance(model, RidgeClassifier):
        param_dist = {
            'classifier__alpha': np.logspace(-4, 4, 20),
            'classifier__solver': ['auto', 'svd', 'cholesky', 'lsqr']
        }
    elif isinstance(model, LinearSVC):
        param_dist = {
            'classifier__C': np.logspace(-4, 4, 20),
            'classifier__loss': ['hinge', 'squared_hinge']
        }
    elif isinstance(model, KNeighborsClassifier):
        param_dist = {
            'classifier__n_neighbors': np.arange(1, 31),
            'classifier__weights': ['uniform', 'distance'],
            'classifier__metric': ['euclidean', 'manhattan', 'minkowski'],
        }
    else:
        print(f"Unsupported model type for tuning: {type(model)}")
        return
    
    steps = [('scaler', StandardScaler())]
    if apply_smote:
        steps.append(('smote', SMOTE(random_state=42)))

    base_clf = clone(model)
    steps.append(('classifier', base_clf))
    pipeline = ImbPipeline(steps)

    grid_search = RandomizedSearchCV(
        estimator=pipeline,
        param_distributions=param_dist,
        n_iter=50,
        scoring='roc_auc',
        cv=3,
        random_state=42,
        n_jobs=-1
    )
    grid_search.fit(x_train, y_train)

    tune_pipeline = grid_search.best_estimator_
    best_model = tune_pipeline.named_steps['classifier']

    if hasattr(best_model, 'predict_proba'):
        proba = tune_pipeline.predict_proba(x_test)[:, 1]
    else:
        try:
            proba = tune_pipeline.decision_function(x_test)
            proba = (proba - proba.min()) / (proba.max() - proba.min())
        except:
            proba = tune_pipeline.predict(x_test)


    thresholds = np.linspace(0.1, 0.9, 81)
    best_fbeta, best_thresh = 0, 0.3
    for thresh in thresholds:
        preds = (proba >= thresh).astype(int)
        fbeta = fbeta_score(y_test, preds, beta=2)
        if fbeta > best_fbeta:
            best_fbeta = fbeta
            best_thresh = thresh

    final_preds = (proba >= best_thresh).astype(int)

    report = classification_report(y_test, final_preds, output_dict=True)
    class_metrics = report['1']
    roc_auc = roc_auc_score(y_test, proba)
    precision, recall, _ = precision_recall_curve(y_test, proba)
    pr_auc = auc(recall, precision)

    result = {   
        'model': name,
        'best_threshold': best_thresh,
        'best_fbeta': best_fbeta,
        'roc_auc': roc_auc,
        'pr_auc': pr_auc,
        'precision': class_metrics['precision'],
        'recall': class_metrics['recall'],
        'f1_score': class_metrics['f1-score'],
        'best_params': best_model.get_params(),
        'probs': proba,
        'preds': final_preds,
    }
    all_model_results.append(result)

    print("\n Tuned Model Performance:")
    print(classification_report(y_test, final_preds))
    print("Confusion Matrix:")
    print(confusion_matrix(y_test, final_preds))

    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    RocCurveDisplay.from_estimator(tune_pipeline, x_test, y_test, ax=axes[0])
    axes[0].set_title(f'ROC Curve - {result["model"]}')
    axes[0].grid(True)


    PrecisionRecallDisplay.from_estimator(tune_pipeline, x_test, y_test, ax=axes[1])
    axes[1].set_title(f'Precision-Recall Curve - {result["model"]}')
    axes[1].grid(True)

    plt.tight_layout()
    plt.show()

for i in range(2):
    model_info = all_model_results[i]
    base_model_name = model_info['Model']
    model_class = None
    if 'Logistic' in base_model_name:
        model_class = LogisticRegression()
    elif 'Ridge' in base_model_name:
        model_class = RidgeClassifier()
    elif 'SVM' in base_model_name:
        model_class = LinearSVC()
    elif 'KNN' in base_model_name:
        model_class = KNeighborsClassifier()
        
    model_info['Model_instance'] = model_class

    apply_smote = 'SMOTE' in model_info['Model']
    tune_and_evaluate(model_info, x_train, y_train, x_test, y_test, apply_smote=apply_smote)
```

```python
@ignore_warnings(category=ConvergenceWarning)
def tune_and_evaluate(base_model_info, x_train, y_train, x_test, y_test, apply_smote=False):
    name = base_model_info['Model']
    model = base_model_info['model_instance']
    print(f"\n==Hyperparameter Tuning {name} ===")


    if isinstance(model, DecisionTreeClassifier):
        param_dist = {
            'classifier__max_depth': [None, 5, 10, 20],
            'classifier__min_samples_split': [2, 5, 10],
            'classifier__min_samples_leaf': [1, 2, 4],
            'classifier__criterion': ['gini', 'entropy']
        }
        user_scaler = False
    elif isinstance(model, RandomForestClassifier):
        param_dist = {
            'classifier__n_estimators': [100, 200],
            'classifier__max_depth': [None, 10, 20],
            'classifier__min_samples_split': [2, 5],
            'classifier__min_samples_leaf': [1, 2],
            'classifier__criterion': ['gini', 'entropy']
        }
        user_scaler = False
    elif isinstance(model, xgb.XGBClassifier):
        param_dist = {
            'classifier__n_estimators': [100, 200],
            'classifier__max_depth': [3, 6, 10],
            'classifier__learning_rate': [0.01, 0.1, 0.2],
            'classifier__subsample': [0.8, 1.0],
            'classifier__colsample_bytree': [0.8, 1.0]
        }
        user_scaler = False
    elif isinstance(model, xgb.XGBClassifier):
        param_dist = {
            'classifier__n_estimators': [100, 200],
            'classifier__max_depth': [3, 6, 10],
            'classifier__learning_rate': [0.01, 0.1, 0.2],
            'classifier__subsample': [0.8, 1.0],
            'classifier__colsample_bytree': [0.8, 1.0]
        }
        user_scaler = False
    elif isinstance(model, LGBMClassifier):
        param_dist = {
            'classifier__n_estimators': [100, 200],
            'classifier__max_depth': [3, 6, 10],
            'classifier__learning_rate': [0.01, 0.1, 0.2],
            'classifier__subsample': [0.8, 1.0],
            'classifier__colsample_bytree': [0.8, 1.0]
        }
        user_scaler = False
    elif isinstance(model, CatBoostclassifier):
        param_dist = {
            'classifier__iterations': [100, 200],
            'classifier__depth': [3, 6, 10],
            'classifier__learning_rate': [0.01, 0.1, 0.2],
            'classifier__l2_leaf_reg': [1, 3, 5]
        }
        user_scaler = False
    else:
        print(f"Unsupported model type for tuning: {type(model)}")
        return
    

    steps = []
    if user_scaler:
        from sklearn.preprocessing import StandardScaler
        steps.append(('scaler', StandardScaler()))
    if apply_smote:
        steps.append(('smote', SMOTE(random_state=42)))
    

    base_clf = clone(model)
    steps.append(('classifier', base_clf))
    pipeline = ImbPipeline(steps)


    grid_serach = RandomizedSearchCV(
        estimator=pipeline,
        param_distributions={'clf__' + key: value for key, value in param_dist.items()},
        n_iter=20,
        scoring='f1',
        cv=3,
        verbose=1,
        random_state=42,
        n_jobs=-1
    )
    grid_serach.fit(x_train, y_train)

    tuned_pipeline = grid_serach.best_estimator_
    best_model = tuned_pipeline.named_steps['classifier']

    if hasattr(best_model, 'predict_proba'):
        proba = tuned_pipeline.predict_proba(x_test)[:, 1]
    else:
        try:
            proba = tuned_pipeline.decision_function(x_test)
            proba = (proba - proba.min()) / (proba.max() - proba.min())
        except:
            proba = tuned_pipeline.predict(x_test)

    thresholds = np.linspace(0.1, 0.9, 81)
    best_fbeta, best_thresh = 0, 0.3
    for thresh in thresholds:
        preds = (proba >= thresh).astype(int)
        fbeta = fbeta_score(y_test, preds, beta=0.5)
        if fbeta > best_fbeta:
            best_fbeta = fbeta
            best_thresh = thresh
    
    final_preds = (proba >= best_thresh).astype(int)

    report = classification_report(y_test, final_preds, output_dict=True)
    class_metrics = report['1']
    roc_auc = roc_auc_score(y_test, proba)
    precision, recall, _ = precision_recall_curve(y_test, proba)
    pr_auc = auc(recall, precision)

    result = {
        'Model': name,
        'model_instance': best_model,
        'best_threshold': best_thresh,
        'roc_auc': roc_auc,
        'pr_auc': pr_auc,
        'precision': class_metrics['precision'],
        'recall': class_metrics['recall'],
        'f1_score': class_metrics['f1-score'],
        'params': grid_serach.best_params_,
        'probs': proba,
        'preds': final_preds
    }
    all_model_results.append(result)


    print(f"\nTuned Model Report:")
    print(classification_report(y_test, final_preds))
    print("Confusion Matrix:")
    print(confusion_matrix(y_test, final_preds))


    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    RocCurveDisplay.from_estimator(tuned_pipeline, x_test, y_test, ax=axes[0])
    axes[0].set_title(f'ROC Curve: {result["Model"]}')
    axes[0].grid(True)

    PrecisionRecallDisplay.from_estimator(y_test, proba, ax=axes[1])
    axes[1].set_title(f'Precision-Recall Curve: {result["Model"]}')
    axes[1].grid(True)

    plt.tight_layout()
    plt.show()



    for i in range(2):
        model_info = all_model_results[i]
        base_model_name = model_info['Model']
        model_instance = None

    if 'Decision Tree' in base_model_name:
        model_instance = DecisionTreeClassifier(random_state=42, class_weight='balanced')
    elif 'Random Forest' in base_model_name:
        model_instance = RandomForestClassifier(random_state=42, class_weight='balanced')
    elif 'XGBoost' in base_model_name:
        model_instance = xgb.XGBClassifier(random_state=42, class_weight='balanced')
    elif 'LightGBM' in base_model_name:
        model_instance = LGBMClassifier(random_state=42, class_weight='balanced')
    elif 'CatBoost' in base_model_name:
        model_instance = CatBoostClassifier(random_state=42, class_weight='balanced', verbose=0)

    
    model_info['model_instance'] = model_instance

    apply_smote = 'Smote' in model_info['Model']
```
